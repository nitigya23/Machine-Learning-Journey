```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
webpage=requests.get("https://quotes.toscrape.com/").text
```

```python
soup=BeautifulSoup(webpage,'lxml')
```

```python
print(soup.prettify())
```

```python
soup.find_all('div',class_='quote')
```

```python
quotes=soup.find_all('div',class_="quote")
```

```python
quote=[]
authors=[]
tags=[]
for i in quotes:
  quote.append(i.find('span',class_='text').text.strip())
  authors.append(i.find('small',class_='author').text.strip())
  tags.append(i.find('div',class_='tags').text.strip('\n'))
```

```python
quote
```

```python
authors
```

```python
tags
```

```python
d={'Quote':quote,'Author':authors,'Tags':tags}
df=pd.DataFrame(d)
```

```python
df.to_csv('quotes.csv')
```

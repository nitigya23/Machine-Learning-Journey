### Web Scraping using `BeautifulSoup` from `bs4`

```python
import pandas as pd
import requests  #used for http request
from bs4 import BeautifulSoup # used as HTML parser & webscraping
```

```python
headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
webpage=requests.get('https://www.ambitionbox.com/list-of-companies').text
```

```python
soup=BeautifulSoup(webpage,'lxml') #lxml is the parse which will parse the webpage
```

```python
print(soup.prettify())  #this will actually print the html formatted data available on that page
```

```python
for i in soup.find_all('h1'):
  print(i.text.strip())       #this will find out all the h1 headings from the webpage


  #all the headings will be in a well formatted structure because of `strip()` function
  # this strip() will remove all the gaps among text

```

### <p>in beautiful soup ,
`h1,h2` tags are used to acces the headings from the webpage
else everything is accessed using `p` tag!</p>

since every paragraph in html have same `classes/div id's` therefore we can directly access them using their `class` name or `div id's` name


```python
len(soup.find_all('p',class_='rating'))   #if length matche sthe list of companies then its right!
```

```python
company=soup.find_all('div',class_='company-content-wrapper')
```

```python
name=[]
rating=[]
reviews=[]
ctype=[]
hq=[]
old=[]
employees=[]
for i in company:    #reading data inside one div all at once (one div at the webpage)
  name.append(i.find('h2').text.strip())
  rating.append(i.find('p',class_='rating').text.strip())
  reviews.append(i.find('a',class_='review-count'))
  ctype.append(i.find_all('p',class_='infoEntity')[0].text.strip())
  hq.append(i.find_all('p',class_='infoEntity')[1].text.strip())
  old.append(i.find_all('p',class_='infoEntity')[2].text.strip())
  employees.append(i.find_all('p',class_='infoEntity')[3].text.strip())

  d={'name':name,'rating':rating,'review':review,'type':ctype,'hq':hq,'old':old,'employees':employees}
  df=pd.DataFrame(d)

```

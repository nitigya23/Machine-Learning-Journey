### while performing `EDA` Exploratory data Analysis, we need to simply perform analysis on  `Categorical data` before `Numerical data`


while Univariate analysis our main focus should always be plotting more and more graphs so that we can analyse and visulaize data really well!

## Categorical data

### 1) Countplot

```python
# countplot
# sns.countplot(df['Survived'])
df['Survived'].value_counts().plot(kind='bar')
```

```python
sns.countplot(x=df['Pclass'])
```

```python
sns.countplot(df['Sex'])
```

```python
sns.countplot(df['Embarked'])  #here S,Q,C reprsent different towns name
#from where the passengers started off their journey!
```

### 2) PieChart

```python
df['Survived'].value_counts().plot(kind='pie',autopct='%.2f')  #autopct provides us the percentage
# value also!
```

```python
df['Pclass'].value_counts().plot(kind="pie",autopct="%.2f")
```

## Numerical Data

### 1) Histogram
#### used to plot frequency distribution of any data

```python
import matplotlib.pyplot as plt
plt.hist(df['Age'],bins=14)
```

### 2) Distplot

->used to plot frequency distribution
->also used in finding Skweness of data-which side data is more skewed!

```python
sns.distplot(df['Age'])  # prints pdf curve-probability distribution curve
```

### 3) Boxplot
#### Prints a 5 number summary i.e
min
Q1
Q2 (median)
Q3
max

IQR= inyterquartile range (Q3-Q1)

```python
sns.boxplot(df['Fare'])
```

```python
sns.boxplot(df['Age'])
```

> ## Sometimes the data we are working with contains numerical as well as categorical data in the same column and we need to split that column into two seperate columns `col_numerical`   and `col_categorical` for better understanding .
this is a part of feature engg.



```python
import numpy as np
import pandas as pd
```

```python
df= pd.read_csv("titanic.csv")
```

```python
df.head()
```

```python
df['number'].unique()   # number column represents how many persons that passenger was travelling with
```

```python
fig = df['number'].value_counts().plot.bar()
fig.set_title("passenger travellinng with")
```

```python
#extract numerical part
df['number_numerical'] = pd.to_numeric(df['number'],errors='coerce',downcast='integer')
```

```python
#extracr catewgorical part
df['number_categorical']= np.where(df['number_numerical'].isnull(),df['number'],np.nan)
df.head()
```

```python
df['Cabin'].unique()
```

```python
df['Ticket'].unique()
```

```python
df['cabin_num']=df['Cabin'].str.extract('(\d+)') #captures numerical part)
df['cabin_cat']= df['Cabin'].str[0]  #captures the first letter

df.head()
```

```python
df['cabin_cat'].value_counts().plot(kind='bar')
```

```python
#extract the last bit of tickdt as number
df['Ticket_num']= df['Ticket'].apply(lambda s:s.split()[-1])
df['Ticket_num']= pd.to_numeric(df['Ticket_num'],
                                errors= 'coerce',
                                downcast='integer')

#extract the first part of ticket as category

df['Ticket_cat']= df['Ticket'].apply(lambda s:s.split()[0])
df['Ticket_cat']= np.where(df['Ticket_cat'].str.isdigit(),
                           np.nan, df['Ticket_cat'])


df.head(5)
```

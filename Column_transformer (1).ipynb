### Why Column Transformer??
till now, we had to apply different encoding techniques to our data because each column had different problems. that's why we  are now using `column transformer` which is a class of `scikit learn` library..

> through this we can literally apply diff. techniques on diff cols depening upon nature of categorical data (ordinal or nominal)

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder
```

 >`Transformers` are list of tuples -> we can have different number of tuples inside transformer since we need to apply encoding technique on 3 different columns that's why we will have to pass `3 tuples`.


> `Remainders` -> what to do with the remaining columns?

> we can do two things:
1.    passthrough -> leave them as it is
2.   drop-> delete them from the dataset



```python
transformer = ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(),['fever']),
 ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
  ('tnf3',OneHotEncoder(drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test).shape
```

```python
#we will do this entrie process in loop
# for this we will create an empty dataframe and will apply a loop
# then keep on appending the movie data one by one
```

```python
df=pd.DataFrame()
```

```python
all_dfs = []
for i in range(400):
  res=requests.get(f'https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={i+1}')
  temp_df=pd.DataFrame(res.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
  all_dfs.append(temp_df)

df = pd.concat(all_dfs, ignore_index=True)
#ignore index will create a new index ignoring the previous one
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df= pd.read_csv("placement.csv")
```

```python
df.head()
```

## Plotting the dataset we are having:

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package')
plt.show()
```

```python
x=df.iloc[:,0:1]
y= df.iloc[:,-1]
```

```python
y
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
x_test
```

```python
lr.predict(x_test.iloc[0].values.reshape(1,1))
```

## output came out as 3.89 but actually the 0th index value of package is 4.10 so our model is a bit off!

```python
lr.predict(x_test.iloc[1].values.reshape(1,1))
```

# Looking how the model actually draws the best fit line for our dataset.

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(x_test,lr.predict(x_test),color='red')
```
